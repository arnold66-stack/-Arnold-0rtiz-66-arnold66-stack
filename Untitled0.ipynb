Analisis de daos utilizando las librerias Numpy, Pandas y Matplolib

```python
import pandas as pd
datos ={
    "ESTUDIATES": ["ANA", "STIVEN", "JUAN", "ARNOLD", "LUISA"],
    "ASISTENCIA": [9,34,55,45,87],
    "TAREAS": [5,6,12,4,7],
    "CALIFICACION": [4.3,5.0,1.5,2.5,3.0]
}
df=pd.DataFrame(datos)
print(df)
#df.head()
#df.info()
#df.describe()
#Promedio de calificaciones
prom = df['CALIFICACION'].mean()
print("El promedio de calificaciones es:",prom)
df["CALIFICACION"].max()
#Ordenar los datos de menor a mayor
df.sort_values('CALIFICACION')
#Ordenar datos de calificacion de mayor a menor
#df.sort_values('CALIFICACION', ascending=0)

#alertas programing
alertas=df[
    (df['ASISTENCIA']<80)&
    (df['CALIFICACION']<3.0)
]
print(alertas)

import matplotlib.pyplot as plt
plt.bar(
    df['ESTUDIATES'],
    df['CALIFICACION']
)
plt.title("Calificaciones estudianes")
plt.xlabel("ESTUDIANTES")
plt.ylabel("CALIFICACIONES")
plt.show()

#Segudo grafico
plt.scatter(
    df['ASISTENCIA'],
    df['CALIFICACION']
)
plt.title("Relacion entre estudiantes y calificación")
plt.xlabel("Asistencia (%)")
plt.ylabel("Calificacion")
plt.show()
```
